## 2. Kiểm Tra Cấu Hình Workflow CI/CD (`ci-aspx-481.yml`)

Đối với project ASP.NET Web Forms (.NET Framework 4.8.1), runner bắt buộc phải là `windows-latest` vì cần bộ công cụ MSBuild bản đầy đủ và .NET Framework SDK. Dưới đây là đoạn code kiểm tra tính hợp lệ của workflow YAML.

```python
import yaml

yaml_content = """
name: CI ASPX 4.8.1
on:
  push:
    branches: [ main, develop ]
  pull_request:
    branches: [ main ]
jobs:
  build:
    runs-on: windows-latest
    steps:
      - name: Checkout
        uses: actions/checkout@v4
      - name: Setup MSBuild
        uses: microsoft/setup-msbuild@v2
      - name: Setup NuGet
        uses: nuget/setup-nuget@v2
      - name: Restore Packages
        run: nuget restore AspxDemo_NET481.sln
      - name: Build Release
        run: msbuild AspxDemo_NET481.sln /p:Configuration=Release
"""

parsed = yaml.safe_load(yaml_content)
print("Tên Workflow:", parsed["name"])
print("Triggers kích hoạt:", list(parsed["on"].keys()))
print("Runner yêu cầu:", parsed["jobs"]["build"]["runs-on"])
assert parsed["jobs"]["build"]["runs-on"] == "windows-latest", "LỖI: Bắt buộc dùng windows-latest cho ASPX 4.8.1!"
print("==> ĐÃ XÁC THỰC: Cấu hình workflow hoàn toàn chuẩn xác cho ASP.NET 4.8.1!")
```

## 3. Mô Phỏng Kiểm Tra An Ninh Mã Nguồn (Secret Scanning)

**Quy tắc vàng:** Tuyệt đối không commit chuỗi kết nối chứa mật khẩu cơ sở dữ liệu (`Web.config`) lên GitHub. Dưới đây là hàm kiểm tra phát hiện lộ secret tự động.

```python
import re

def scan_secrets(config_text):
    patterns = [
        r"password\s*=\s*[^;\"'>]+",
        r"User ID\s*=\s*sa;",
        r"Initial Catalog\s*=\s*[^;\"'>]+"
    ]
    found = []
    for p in patterns:
        matches = re.findall(p, config_text, re.IGNORECASE)
        if matches:
            found.extend(matches)
    return found

bad_config = '<connectionStrings><add name="Sql" connectionString="Data Source=.\\SQLEXPRESS;Initial Catalog=QLSV;User ID=sa;Password=Secret123!"/></connectionStrings>'
good_config = '<connectionStrings><add name="Sql" connectionString="Data Source=localhost;Initial Catalog=QLSV;Integrated Security=True"/></connectionStrings>'

print("Quét cấu hình không an toàn:", scan_secrets(bad_config))
print("Quét cấu hình an toàn:", scan_secrets(good_config))
```
